```python
pirate_agent = Agent(
    "groq:llama-3.3-70b-versatile",
    description="An AI agent with a pirate personna",
    system_prompt=(
        "You are Captain Monkey D Luffy. Who is on a mission to find one pice"
        "You must answer every question like a king of pirates"
        "Never drop your character."
    )
)

res = pirate_agent.run_sync("Explain how get food on a ship. do u like nami, and how do u view zoro like what are your thoughts on him")

print("Pirate Agent Ans: ", res.output)
```

```python
# Runs Async

translator_agent = Agent(
    "groq:openai/gpt-oss-120b",
    system_prompt="Translate the given text to Korean"
)

translator_agent
```

```python
async def get_async_translation():
    print("Sending request to AI...")
    # your computer can do other works    
    res = await translator_agent.run("hello, how are you. My name is Prateek")
    print("Async translate: ", res.output)
    
await get_async_translation()
    
```

```python
story_agent=Agent(
    "groq:llama-3.3-70b-versatile",
    system_prompt="Write a creative 10-sentence story."
)

async def stream_story():
    print("Streaming Response:\n")
    
    async with story_agent.run_stream("Tell me about a brave solder.") as result:
        async for word_chunk in result.stream_text(delta=True):
            print(word_chunk, end="", flush=True)
    
    print()
    
await stream_story()
```
